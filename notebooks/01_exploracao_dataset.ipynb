```python
from pathlib import Path
import yaml
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

# Configurações de exibição
plt.style.use('seaborn-v0_8-whitegrid')
DATA_DIR = Path("../data/garbage-classification-3")
print(f"Caminho do Dataset: {DATA_DIR.resolve()}")
```

## 1. Leitura do arquivo `data.yaml`

```python
yaml_path = DATA_DIR / "data.yaml"
if yaml_path.exists():
    with open(yaml_path, "r") as f:
        data_config = yaml.safe_load(f)
    print("Classes registradas:", data_config.get("names"))
    print("Número de classes:", data_config.get("nc"))
else:
    print("Arquivo data.yaml não encontrado. Baixe o dataset primeiro executando o script download_dataset.py.")
```

## 2. Contagem de Imagens e Distribuição de Classes

```python
splits = ["train", "valid", "test"]
counts = {}
for split in splits:
    images_path = DATA_DIR / split / "images"
    if images_path.exists():
        imgs = list(images_path.glob("*.jpg")) + list(images_path.glob("*.png"))
        counts[split] = len(imgs)
    else:
        counts[split] = 0

df_counts = pd.DataFrame(list(counts.items()), columns=["Split", "Total de Imagens"])
df_counts
```
